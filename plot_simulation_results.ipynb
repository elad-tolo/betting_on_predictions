# Plot Multivariate Results

```python
import os
from collections import defaultdict

data_dir = "results/real_data/multivariate_health_care"
pkl_files = [f for f in os.listdir(data_dir) if f.endswith('.pkl') and 'seed' in f and 'validity_False' in f]

statistic_to_rejections = defaultdict(list)

# Load and merge all files
for pkl_file in sorted(pkl_files):
    file_path = os.path.join(data_dir, pkl_file)
    with open(file_path, "rb") as f:
        data = pickle.load(f)
        for key, values in data["rejection_lists"].items():
            statistic_to_rejections[key] += values

p, ax = plot_power_vs_steps_paper(
    statistic_to_rejections,
    show_legend=True,
    legend_font_size=15,
    save_path="plots/real_data/multivariate_health_care.pdf",
)
```
